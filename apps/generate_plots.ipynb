```python
import pickle

from pinn_jc.io.legacy import load_legacy_pickle
import numpy as np

from pinn_jc.visualization.plots import (
    plot_expected_values,
    plot_states,
    plot_loss_functions,
    plot_learned_param,
    plot_fidelity
)
```

```python
params = {"wc": 1.0, "wa": 1.0, "g1": 1}
tfinal = 2 * np.pi
n_time_steps = 200
picture = "atom"
epochs = 30000
n_points_loss = n_time_steps
```

## Fock state

```python
init_state = "fock"

dims = {"atom": 2, "field": 2}

results_dict = load_legacy_pickle(f"../saved_artifacts/training/results_dict_{init_state}.pkl")

models_dict = results_dict["run_1"]["models"]
loss_dict = results_dict["run_1"]["loss"]

plot_expected_values(
    models_dict,
    tfinal,
    n_time_steps,
    init_state,
    params,
    picture,
    dims,
    is_scaled=False,
    train_or_test="train",
    plot_input="expected",
)

plot_states(
    models_dict,
    params,
    tfinal,
    init_state=init_state,
    dims=dims,
    n_time_steps=1000,
    picture=picture,
    is_scaled=False,
    train_or_test="train",
    plot_input="state",
)

plot_loss_functions(loss_dict, skip_param=300)

plot_learned_param(loss_dict, true_param=1, skip_param=300, picture=picture)

plot_fidelity(
    models_dict,
    params,
    tfinal,
    init_state=init_state,
    dims=dims,
    n_time_steps=200,
    picture=picture,
    is_scaled=False,
    train_or_test="train",
    plot_input="state",
)
```

## Rabi

```python
init_state = "fock"
picture = "rabi2"

dims = {"atom": 2, "field": 2}
params = {"wc": 10, "wa": 10, "g1":0.1, "g2":0.}
tfinal = 8*np.pi

results_dict = load_legacy_pickle(f"../saved_artifacts/training/results_dict_{picture}_{init_state}.pkl")

for i in range(5):
    models_dict = results_dict[f"run_{i}"]["models"]
    loss_dict = results_dict[f"run_{i}"]["loss"]
    print("Run:", i)
    plot_states(
        models_dict=models_dict,
        params=params,
        tfinal=tfinal,
        init_state=init_state,
        dims=dims,
        n_time_steps=1000,
        picture="interaction",
        is_scaled=False,
        train_or_test="train",
        plot_input="state",
    )
    plot_expected_values(
        models_dict=models_dict,
        tfinal=tfinal,
        n_time_steps=n_time_steps,
        init_state=init_state,
        params=params,
        picture="interaction",
        dims=dims,
        is_scaled=False,
        train_or_test="train",
        plot_input="expected",
    )
    try:
        plot_learned_param(loss_dict=loss_dict, true_param=1, skip_param=500, picture="rabi2")
    except:
        pass
    plot_loss_functions(loss_dict, skip_param=500)

    plot_fidelity(
        models_dict=models_dict,
        params=params,
        tfinal=tfinal,
        init_state=init_state,
        dims=dims,
        n_time_steps=200,
        picture="interaction",
        is_scaled=False,
        train_or_test="train",
        plot_input="state",
    )
```

```python
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 1000)
y = np.log(1+np.exp(x))
dy_dx = 1/(1+np.exp(-x))
plt.figure(dpi=300)
plt.xlabel(r"\(\theta\)")
# plt.ylabel(r"softplus(\(\theta\))")
plt.plot(x, y, label=r"softplus(\(\theta\))")
plt.plot(x, dy_dx, label=r"softplus'(\(\theta\))")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 1000)
y = np.abs(x)
dy_dx1 = np.repeat(-1, len(x)//2)
dy_dx2 = np.repeat(1, len(x)//2)
plt.figure(dpi=300)
plt.xlabel(r"\(\theta\)")
plt.plot(x, y, label=r"\(|\theta|\)")
plt.plot(x[:len(x)//2], dy_dx1, label=r"\(|\theta|\)'", color='orange')
plt.plot(x[(len(x)//2):], dy_dx2, color='orange')
plt.legend()
plt.show()
```
